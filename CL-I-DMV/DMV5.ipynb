```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Read dataset
df = pd.read_csv("City_Air_Quality.csv")

# 2. Explore dataset
print(df.head())
print(df.info())
print(df.describe())

# Convert Date to datetime
df["Date"] = pd.to_datetime(df["Date"])

# 3 & 4. AQI trend
plt.plot(df["Date"], df["AQI"], label="AQI")
plt.xlabel("Date")
plt.ylabel("AQI")
plt.title("AQI Trend Over Time")
plt.legend()
plt.xticks(rotation=45)
plt.show()

# 5. Pollutant trends
plt.plot(df["Date"], df["PM2.5"], label="PM2.5")
plt.plot(df["Date"], df["PM10"], label="PM10")
plt.plot(df["Date"], df["CO"], label="CO")
plt.xlabel("Date")
plt.ylabel("Pollutant Level")
plt.title("Pollutant Trends")
plt.legend()
plt.xticks(rotation=45)
plt.show()

# 6. Bar plot of AQI
plt.bar(df["Date"], df["AQI"])
plt.xlabel("Date")
plt.ylabel("AQI")
plt.title("AQI by Date")
plt.xticks(rotation=45)
plt.show()

# 7. Box plot
plt.boxplot(df["AQI"])
plt.ylabel("AQI")
plt.title("AQI Distribution")
plt.show()

# 8. Scatter plot
plt.scatter(df["PM2.5"], df["AQI"])
plt.xlabel("PM2.5")
plt.ylabel("AQI")
plt.title("PM2.5 vs AQI")
plt.show()
```
