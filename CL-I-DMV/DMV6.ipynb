```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Read dataset
df = pd.read_csv("Retail_Sales_Data.csv")

# 2. Explore dataset
print(df.head())
print(df.info())
print(df.describe())

# 3 & 4. Total sales by product category
sales = df.groupby("Product Category")["Total Amount"].sum()

print("\nSales by Product Category:")
print(sales)

# 5. Bar plot
sales.plot(kind="bar")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.title("Sales by Product Category")
plt.show()

# 6. Top-performing category
print("\nTop Performing Category:")
print(sales.idxmax())
print("Sales:", sales.max())

# 7. Sales by Gender and Product Category
category_sales = df.groupby(
    ["Gender", "Product Category"]
)["Total Amount"].sum()

print("\nSales by Gender and Product Category:")
print(category_sales)

# 8. Grouped bar plot
category_sales.unstack().plot(kind="bar")
plt.xlabel("Gender")
plt.ylabel("Total Sales")
plt.title("Sales by Gender and Product Category")
plt.show()
```
