```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
url = "https://media.geeksforgeeks.org/wp-content/uploads/Wine.csv"
df = pd.read_csv(url)
```

```python
X = df.drop('Customer_Segment', axis=1)
y = df['Customer_Segment']
```

```python
y_binary = y.map({1: 'Red Wine', 2: 'White Wine', 3: 'White Wine'})
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
```

```python
print(f"Variance explained by PC1: {pca.explained_variance_ratio_[0]:.2%}")
print(f"Variance explained by PC2: {pca.explained_variance_ratio_[1]:.2%}")
print(
    f"Total Variance Captured: {sum(pca.explained_variance_ratio_):.2%}"
)
```

```python
plt.figure(figsize=(8, 6))
for wine_type in ['Red Wine', 'White Wine']:
    plt.scatter(
        X_pca[y_binary == wine_type, 0], 
        X_pca[y_binary == wine_type, 1], 
        label=wine_type
    )
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA: 2 Clusters (Red Wine vs White Wine)')
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Read dataset
df = pd.read_csv("https://media.geeksforgeeks.org/wp-content/uploads/Wine.csv")

X = df.drop("Customer_Segment", axis=1)
y = df["Customer_Segment"]

# Convert 3 classes into 2 components
y_binary = y.map({1: "Red Wine", 2: "White Wine"})

# Scale data
X_scaled = StandardScaler().fit_transform(X)

# Apply PCA with 2 components
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Display explained variance
print("PC1 Variance:", round(pca.explained_variance_ratio_[0] * 100, 2), "%")
print("PC2 Variance:", round(pca.explained_variance_ratio_[1] * 100, 2), "%")
print("Total Variance:", round(sum(pca.explained_variance_ratio_) * 100, 2), "%")

# Plot using two colours
for wine in ["Red Wine", "White Wine"]:
    plt.scatter(
        X_pca[y_binary == wine, 0],
        X_pca[y_binary == wine, 1],
        label=wine
    )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA: Red Wine vs White Wine")
plt.legend()
plt.show()
```
