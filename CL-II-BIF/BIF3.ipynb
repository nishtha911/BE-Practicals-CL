```python
from vina import Vina
import pandas as pd
import matplotlib.pyplot as plt

protein = "protein.pdbqt"
ligands = ["ligand1.pdbqt", "ligand2.pdbqt", "ligand3.pdbqt"]

results = []

for ligand in ligands:
    v = Vina(sf_name="vina")
    v.set_receptor(protein)
    v.set_ligand_from_file(ligand)

    v.compute_vina_maps(
        center=[10, 10, 10],
        box_size=[20, 20, 20]
    )

    v.dock(exhaustiveness=8, n_poses=1)

    score = v.energies()[0][0]
    results.append([ligand, score])

results = pd.DataFrame(
    results,
    columns=["Ligand", "Binding Affinity (kcal/mol)"]
)

results = results.sort_values(
    "Binding Affinity (kcal/mol)"
)

print(results)

results.plot(
    x="Ligand",
    y="Binding Affinity (kcal/mol)",
    kind="bar",
    legend=False
)

plt.ylabel("Binding Affinity (kcal/mol)")
plt.title("Virtual Screening Results")
plt.show()
```
