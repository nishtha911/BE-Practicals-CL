```python
import nltk
import re

from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

# Download required NLTK resources
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")

# Input text
text = """Natural Language Processing is an important part of Artificial Intelligence.
It helps computers understand and process human languages!"""

print("Original Text:")
print(text)

# 1. Convert to lowercase
text = text.lower()

print("\nAfter Lowercase Conversion:")
print(text)

# 2. Remove punctuation and special characters
text = re.sub(r"[^a-zA-Z\s]", "", text)

print("\nAfter Removing Punctuation:")
print(text)

# 3. Tokenization
words = word_tokenize(text)

print("\nAfter Tokenization:")
print(words)

# 4. Remove stop words
stop_words = set(stopwords.words("english"))

filtered_words = []

for word in words:
    if word not in stop_words:
        filtered_words.append(word)

print("\nAfter Stop Word Removal:")
print(filtered_words)

# 5. Stemming
stemmer = PorterStemmer()

stemmed_words = []

for word in filtered_words:
    stemmed_words.append(stemmer.stem(word))

print("\nAfter Stemming:")
print(stemmed_words)

# Final preprocessed text
processed_text = " ".join(stemmed_words)

print("\nFinal Preprocessed Text:")
print(processed_text)
```
